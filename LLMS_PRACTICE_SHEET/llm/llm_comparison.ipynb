```python
from google import genai
from groq import Groq
from dotenv import load_dotenv

import os
import pandas as pd
```

```python
load_dotenv()

gemini_key = os.getenv("GEMINI_API_KEY")
groq_key = os.getenv("GROQ_API_KEY")

print("Gemini Key Loaded:", gemini_key is not None)
print("Groq Key Loaded:", groq_key is not None)
```

```python
gemini_client = genai.Client(
    api_key=gemini_key
)

groq_client = Groq(
    api_key=groq_key
)

print("Clients Initialized Successfully")
```

```python
prompt = """
Explain Machine Learning in simple words.
"""
```

```python
gemini_response = gemini_client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

gemini_text = gemini_response.text

print(gemini_text)
```

```python
from groq import Groq

groq_client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

response = groq_client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {
            "role": "user",
            "content": "Explain Machine Learning in simple words."
        }
    ]
)

print(response.choices[0].message.content)
```

```python
groq_client = Groq(
    api_key=groq_key
)

groq_response = groq_client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

groq_text = groq_response.choices[0].message.content

print(groq_text)
```

```python
comparison = pd.DataFrame({
    "Model": ["Gemini", "Groq Llama"],
    "Response": [gemini_text, groq_text]
})

comparison
```

```python
comparison.to_csv(
    "responses.csv",
    index=False
)

print("Responses saved successfully")
```

# Multi-LLM Comparison Observations

Both Gemini and Groq successfully generated responses for the same prompt.

Gemini provided a more detailed and educational explanation with examples.

Groq generated a concise and straightforward response.

This experiment demonstrates that different Large Language Models can produce different response styles, levels of detail, and presentation formats even when given the same prompt.
