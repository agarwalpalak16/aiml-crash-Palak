```python
import os
import pandas as pd

from dotenv import load_dotenv
from google import genai
from groq import Groq
```

```python
load_dotenv()

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
GROQ_API_KEY = os.getenv("GROQ_API_KEY")

print("Gemini:", GEMINI_API_KEY is not None)
print("Groq:", GROQ_API_KEY is not None)
```

```python
gemini_client = genai.Client(
    api_key=GEMINI_API_KEY
)

groq_client = Groq(
    api_key=GROQ_API_KEY
)

print("Clients Ready")
```

```python
prompts = {
    "Basic":
        "Explain Machine Learning.",

    "Detailed":
        "Explain Machine Learning in detail with examples.",

    "Beginner":
        "Explain Machine Learning as if teaching a 10-year-old child.",

    "Professional":
        "Explain Machine Learning for a software engineering interview."
}
```

```python
gemini_results = []

for prompt_type, prompt in prompts.items():

    try:
        response = gemini_client.models.generate_content(
            model="gemini-2.0-flash",
            contents=prompt
        )

        gemini_results.append([
            prompt_type,
            response.text
        ])

    except Exception as e:

        gemini_results.append([
            prompt_type,
            f"ERROR: {e}"
        ])

print("Gemini Completed")
```

Gemini API requests were temporarily limited due to quota restrictions during experimentation. Groq API was used successfully for prompt engineering analysis.

```python
gemini_df = pd.DataFrame(
    gemini_results,
    columns=["Prompt Type", "Response"]
)

gemini_df
```

```python
groq_results = []

for prompt_type, prompt in prompts.items():

    response = groq_client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    groq_results.append([
        prompt_type,
        response.choices[0].message.content
    ])

print("Groq Completed")
```

```python
groq_results = []

for prompt_type, prompt in prompts.items():

    response = groq_client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    groq_results.append([
        prompt_type,
        response.choices[0].message.content
    ])

print("Groq Completed")
```

```python
groq_df = pd.DataFrame(
    groq_results,
    columns=["Prompt Type", "Response"]
)

groq_df
```

```python
gemini_df.to_csv(
    "gemini_prompt_results.csv",
    index=False
)

groq_df.to_csv(
    "groq_prompt_results.csv",
    index=False
)

print("Files Saved")
```
